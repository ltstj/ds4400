```python
#Q5
#packages
import pandas as pd

#loading dataset
df = pd.read_csv('kc_house_data.csv')

#dropping useless columns
df = df.drop(['id', 'date', 'zipcode'], axis=1, errors='ignore')

#features without price
features = [col for col in df.columns if col != 'price']
```

```python
#part 1
#mean, minimum, maximum, variance
stats = pd.DataFrame({
    'mean': df[features].mean(),
    'min': df[features].min(),
    'max': df[features].max(),
    'variance': df[features].var()
})
print("Feature Stats:")
print(stats)

l_mean  = stats['mean'].idxmin(), stats['mean'].min()
h_mean = stats['mean'].idxmax(), stats['mean'].max()
l_var = stats['variance'].idxmin(), stats['variance'].min()
h_var = stats['variance'].idxmax(), stats['variance'].max()

print(f"Lowest average feature: {l_mean[0]} ({l_mean[1]})")
print(f"Highest average feature: {h_mean[0]} ({h_mean[1]})")
print(f"Lowest variance feature: {l_var[0]} ({l_var[1]})")
print(f"Highest variance feature: {h_var[0]} ({h_var[1]})")
```

```python
#part 2
feat_corr = df.corr(numeric_only=True)['price'].drop('price').sort_values(ascending=False)
print("Correlation with sorted price:")
print(feat_corr.to_frame(name='Correlation'))
print("Highest correlation feature:", feat_corr.idxmax(), f"({feat_corr.max()})")
```
